## CLEANING THE DATA

```python
# Remove duplicate records
df = df.drop_duplicates().reset_index(drop=True)

# Fill missing numerical values using the median
numeric_columns = df.select_dtypes(include=np.number).columns

df[numeric_columns] = df[numeric_columns].apply(
    lambda col: col.fillna(col.median())
)

print("Missing values after cleaning:")
print(df.isnull().sum().sum())

print("Final dataset shape:", df.shape)
```

```python
# Display descriptive statistics
df.describe().T
```

```python
# Detect outliers using the IQR method

numeric_columns = df.select_dtypes(include=np.number).columns

outlier_summary = []

for col in numeric_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = ((df[col] < lower_limit) | (df[col] > upper_limit)).sum()

    outlier_summary.append([col, outliers])

outlier_summary = pd.DataFrame(
    outlier_summary,
    columns=["Column", "Outlier_Count"]
)

# Show columns containing outliers
outlier_summary = outlier_summary[
    outlier_summary["Outlier_Count"] > 0
].sort_values("Outlier_Count", ascending=False)

display(outlier_summary)
```

#ANALYSIS

###Seasonal Performance

```python
# Count records in each season
season_count = df["Season"].value_counts()

plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="Season",
    order=season_count.index
)

plt.title("Distribution of Agricultural Records by Season")
plt.xlabel("Season")
plt.ylabel("Number of Records")

plt.show()
```

```python
# Calculate average performance for each season
season_performance = (
    df.groupby("Season")
    .agg(
        Avg_Yield=("Yield_Tonnes_Ha", "mean"),
        Avg_Production=("Production_Tonnes", "mean"),
        Avg_Revenue=("Revenue_INR", "mean"),
        Avg_Profit=("Profit_INR", "mean"),
        Avg_Water_Efficiency=("Water_Efficiency_t_per_1000m3", "mean")
    )
    .round(2)
)

display(season_performance)
```

```python
# Plot seasonal performance
season_performance[
    ["Avg_Yield", "Avg_Profit"]
].plot(
    kind="bar",
    figsize=(10, 6)
)

plt.title("Agricultural Performance Across Seasons")
plt.xlabel("Season")
plt.ylabel("Average Performance")
plt.xticks(rotation=0)
plt.legend(["Average Yield", "Average Profit"])

plt.tight_layout()
plt.show()
```

###CROP-WISE PERFORMANCE

```python
# Compare agricultural performance across crops
crop_performance = (
    df.groupby("Crop")
    .agg(
        Avg_Yield=("Yield_Tonnes_Ha", "mean"),
        Avg_Production=("Production_Tonnes", "mean"),
        Avg_Profit=("Profit_INR", "mean")
    )
    .sort_values("Avg_Yield", ascending=False)
    .round(2)
)

display(crop_performance)
```

```python
# Crop-wise yield visualization
plt.figure(figsize=(10, 6))

sns.barplot(
    data=df,
    x="Crop",
    y="Yield_Tonnes_Ha",
    estimator="mean",
    errorbar=None
)

plt.title("Average Yield by Crop")
plt.xlabel("Crop")
plt.ylabel("Average Yield (Tonnes/Ha)")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

###IRRIGATION & WATER EFFICIENCY

```python
# Compare irrigation methods
irrigation_summary = (
    df.groupby("Irrigation_Method")
    .agg(
        Avg_Yield=("Yield_Tonnes_Ha", "mean"),
        Avg_Water_Used=("Water_Used_m3", "mean"),
        Avg_Water_Efficiency=("Water_Efficiency_t_per_1000m3", "mean")
    )
    .round(2)
)

display(irrigation_summary)
```

###SEASON × CROP COMPARISON

```python
# Calculate average yield by season and crop
season_crop = (
    df.groupby(["Season", "Crop"])["Yield_Tonnes_Ha"]
    .mean()
    .unstack()
    .round(2)
)

display(season_crop)
```

```python
plt.figure(figsize=(11, 6))

sns.heatmap(
    season_crop,
    annot=True,
    fmt=".2f"
)

plt.title("Average Yield by Season and Crop")
plt.xlabel("Crop")
plt.ylabel("Season")

plt.tight_layout()
plt.show()
```

###SEASON × IRRIGATION × WATER EFFICIENCY (comparison)

```python
# Compare irrigation efficiency across seasons
season_irrigation = (
    df.groupby(
        ["Season", "Irrigation_Method"]
    )["Water_Efficiency_t_per_1000m3"]
    .mean()
    .unstack()
    .round(2)
)

display(season_irrigation)
```

```python
plt.figure(figsize=(10, 6))

sns.heatmap(
    season_irrigation,
    annot=True,
    fmt=".2f"
)

plt.title(
    "Water Efficiency by Season and Irrigation Method"
)

plt.xlabel("Irrigation Method")
plt.ylabel("Season")

plt.tight_layout()
plt.show()
```

###FINAL AGRICULTURAL PERFORMANCE SUMMARY

```python
# Final overall performance summary
final_summary = {
    "Best Season by Average Yield":
        df.groupby("Season")["Yield_Tonnes_Ha"].mean().idxmax(),

    "Best Season by Average Profit":
        df.groupby("Season")["Profit_INR"].mean().idxmax(),

    "Best Crop by Average Yield":
        df.groupby("Crop")["Yield_Tonnes_Ha"].mean().idxmax(),

    "Best Crop by Average Profit":
        df.groupby("Crop")["Profit_INR"].mean().idxmax(),

    "Best Irrigation Method by Water Efficiency":
        df.groupby("Irrigation_Method")[
            "Water_Efficiency_t_per_1000m3"
        ].mean().idxmax()
}

for key, value in final_summary.items():
    print(f"{key}: {value}")
```

```python
# Create a final agricultural intelligence table
final_comparison = pd.DataFrame({
    "Season": df["Season"].unique()
})

final_comparison["Avg Yield"] = final_comparison["Season"].map(
    df.groupby("Season")["Yield_Tonnes_Ha"].mean()
)

final_comparison["Avg Profit"] = final_comparison["Season"].map(
    df.groupby("Season")["Profit_INR"].mean()
)

final_comparison["Avg Water Efficiency"] = final_comparison["Season"].map(
    df.groupby("Season")[
        "Water_Efficiency_t_per_1000m3"
    ].mean()
)

final_comparison.round(2)
```
